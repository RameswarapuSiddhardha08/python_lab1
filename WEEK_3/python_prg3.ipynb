```python
# check whether given number prime or not
n=int(input())
for i in range(2,n):
  if(n%i==0):
   print("it is not prime")
   break
else:
  print("yes it is prime")
```

```python
# print prime numbers in between two numbers
a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
   if(i%j==0):
    break
  else:
   print(i,end=" ")
```

```python
# print sum of the digits of the given number
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```

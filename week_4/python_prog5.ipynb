```python
  #program that count the no of vowels in a string
  string = input()
  vowels = 0
  for ch in string:
    if ch in "aeiouAEIOU":
      vowels += 1
  print("number of vowels:",vowels)
```

```python
  #program that count the no of vowels and consonants in a string
  string = input()
  vowels = 0
  consonants = 0
  for ch in string:
    if ch in "aeiouAEIOU":
      vowels += 1
    else:
      consonants += 1
  print("number of vowels:",vowels)
  print("number of consonants",consonants)
```

TUPLE

```python
#OPERATIONS ON TUPLE
t1=(7,8,9,11,45.5,36.6,"Vincenzo")
t2=(4,5,7,12,"Cassenno")
print(t1+t2)
print(t1) #concatenation
print(t2*2)  #repetition
for i in t1:     #iteration
  print(i)
if(2 in t1):      #membership
  print(2 in t1)
else:
  print("no")
print(t1[::-1])       #reverse a list
```

```python
#FUNCTIONS IN TUPLE
print(len(t1))
t4=tuple(int(i) for i in input().split())
print(t4)
print(max(t4))
print(min(t4))
print(sorted(t4))
```

```python
#METHODS IN TUPLE
print(t1.count(7))
print(t1.index("Vincenzo"))
t3=tuple(map(int,input().split()))
print(t3)
print(tuple(zip(t3,t4)))
```

```python
#OPERATIONS ON SETS
#1.Union (|)
#2.Intersection (&)
#3.Difference (-)
#4.Symmetric Difference (^)
a={1,2,3}
b={"Siddu",4,5,6.0,3,2}
print(a&b) #intersection
print(a|b) #union
print(a-b) #difference
print(b-a) #difference
print(a^b) #Symmetric difference
print(a)
print(b)
```

Program 4: Write a python program that find missing number between 1 and n using a set

```python
n=int(input())
L1=list(map(int,input().split()))
L2=[]
for i in range(1,n+1):
  L2.append(i)
p=set(L1)
q=set(L2)
print(q-p)
```

PROGRAM 5: Write a python program that prints only unique elements in a list

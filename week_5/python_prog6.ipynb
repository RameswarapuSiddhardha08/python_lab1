```python
#program that checks whether the given number is palindrome or not using function
def is_palindrome(num):
  original = num
  reverse = 0
  while(num>0):
    digit = num%10
    reverse = reverse*10+digit
    num = num//10
  return original == reverse
number=int(input())
if(is_palindrome(number)):
  print("it is a palindrome")
else:
  print("it is not a palindrome")

```

```python
# write a program to print rerverse of given string using function
def reverse_string(s):
   rev=""
   for i in range(len(s)-1,-1,-1):
    rev=rev+s[i]
   return rev
print(reverse_string("hello"))
```

```python
#function that returns multiple values
def multiple_values(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4,5)
print(f"sum is {res_sum},count is {res_count},the average is{res_avg}")
```
